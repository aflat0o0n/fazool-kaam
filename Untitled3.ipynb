````python
# -*- coding: utf-8 -*-
"""Cell 1: Environment setup — Java, apktool, apk.sh, APKEditor, archive tools."""

import os, sys, subprocess, json, urllib.request
from pathlib import Path

def run_cmd(cmd, check=True, **kwargs):
    print("$", " ".join(map(str, cmd)))
    return subprocess.run(cmd, check=check, **kwargs)

# ── System packages ──────────────────────────────────────────────
run_cmd(["apt-get", "update", "-qq"])
run_cmd([
    "apt-get", "install", "-y", "-qq",
    "openjdk-17-jre-headless",
    "unzip",          # ZIP extraction
    "unrar",          # RAR extraction
    "p7zip-full",     # 7z extraction
    "curl", "git",
], stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL)

# Python packages for archive handling & Google Drive
run_cmd([sys.executable, "-m", "pip", "install", "-q",
         "rarfile", "gdown", "requests"], check=False)

print("Java:")
run_cmd(["java", "-version"], check=False)

# ── apktool 2.9.3 ───────────────────────────────────────────────
APKTOOL_JAR = Path("/usr/local/bin/apktool.jar")
APKTOOL_BIN = Path("/usr/local/bin/apktool")

if not APKTOOL_JAR.exists() or APKTOOL_JAR.stat().st_size < 100_000:
    run_cmd([
        "curl", "-fL", "--retry", "3",
        "-o", str(APKTOOL_JAR),
        "https://github.com/iBotPeaches/Apktool/releases/download/"
        "v2.9.3/apktool_2.9.3.jar"
    ])

APKTOOL_BIN.write_text(
    '#!/bin/bash\nexec java -jar /usr/local/bin/apktool.jar "$@"\n'
)
APKTOOL_BIN.chmod(0o755)

r = subprocess.run(["apktool", "--version"],
                   capture_output=True, text=True)
print("apktool:", r.stdout.strip() or r.stderr.strip())

# ── apk.sh ──────────────────────────────────────────────────────
APKSH_DIR = Path("/content/apk.sh")
if not APKSH_DIR.exists():
    run_cmd(["git", "clone", "--depth=1",
             "https://github.com/ax/apk.sh.git", str(APKSH_DIR)])
(APKSH_DIR / "apk.sh").chmod(0o755)
print("apk.sh:", APKSH_DIR / "apk.sh")

# ── APKEditor (latest release .jar) ────────────────────────────
APKEDITOR = Path("/content/APKEditor.jar")

def download_apkeditor():
    api = ("https://api.github.com/repos/"
           "REAndroid/APKEditor/releases/latest")
    req = urllib.request.Request(api,
                                headers={"User-Agent": "Mozilla/5.0"})
    with urllib.request.urlopen(req, timeout=30) as resp:
        release = json.load(resp)

    jar_assets = [a for a in release.get("assets", [])
                  if a.get("name", "").lower().endswith(".jar")]
    if not jar_assets:
        raise RuntimeError("No .jar asset found on APKEditor releases.")

    asset   = jar_assets[0]
    url     = asset["browser_download_url"]
    print("APKEditor release :", release.get("tag_name"))
    print("APKEditor asset  :", asset["name"])

    req = urllib.request.Request(url,
                                headers={"User-Agent": "Mozilla/5.0"})
    with urllib.request.urlopen(req, timeout=120) as resp, \
         APKEDITOR.open("wb") as f:
        while True:
            chunk = resp.read(1024 * 1024)
            if not chunk:
                break
            f.write(chunk)

if not APKEDITOR.exists() or APKEDITOR.stat().st_size < 100_000:
    if APKEDITOR.exists():
        APKEDITOR.unlink()
    download_apkeditor()

print("APKEditor:", APKEDITOR,
      f"({APKEDITOR.stat().st_size / 1e6:.1f} MB)")
print("\n✅ Setup complete.")
````

````python
# -*- coding: utf-8 -*-
"""Cell 3: Download (or receive upload), decompress archives,
identify the real APK/bundle inside, and set input_apk_path."""

import os, sys, re, shutil, binascii, zipfile, subprocess, tarfile
from pathlib import Path
from urllib.parse import urlparse

try:
    import requests
except ImportError:
    subprocess.run([sys.executable, "-m", "pip", "install", "-q", "requests"],
                   check=False)
    import requests

WORK = Path("/content/work")
WORK.mkdir(parents=True, exist_ok=True)
os.chdir(WORK)

# ═══════════════════════════════════════════════════════════════
#  STEP 0 — Resolve the source file
# ═══════════════════════════════════════════════════════════════

if INPUT_MODE == "upload":
    # ── Colab file picker ──────────────────────────────────────
    try:
        from google.colab import files as colab_files
    except ImportError:
        raise RuntimeError(
            "Upload mode requires Google Colab. "
            "Use 'url' or 'local' mode instead."
        )
    print("Opening file picker …")
    uploaded = colab_files.upload()
    if not uploaded:
        raise RuntimeError("No file was uploaded.")
    source_files = []
    for fname, data in uploaded.items():
        dest = WORK / re.sub(r"[^A-Za-z0-9._-]", "_", fname)
        dest.write_bytes(data)
        source_files.append(dest)
        print(f"  ✓ Saved: {dest}  ({len(data)/1e6:.1f} MB)")
    source_path = source_files[0]
    if len(source_files) > 1:
        print(f"  Note: {len(source_files)} files uploaded — "
              f"using first: {source_path.name}")

elif INPUT_MODE == "local":
    # ── Copy local file into WORK ─────────────────────────────
    source_path = Path(LOCAL_PATH.strip())
    if not source_path.exists():
        raise FileNotFoundError(f"File not found: {source_path}")
    dest = WORK / source_path.name
    shutil.copy2(source_path, dest)
    source_path = dest
    print(f"Copied: {source_path}")

elif INPUT_MODE == "url":
    # ── Download from URL ──────────────────────────────────────
    url = DOWNLOAD_URL.strip()
    if not url:
        raise ValueError("DOWNLOAD_URL is empty. Set it in Cell 2.")

    # Output filename
    if BUNDLE_NAME.strip():
        out_name = BUNDLE_NAME.strip()
    else:
        out_name = Path(urlparse(url).path).name or "download.bin"
    out_name = re.sub(r"[^A-Za-z0-9._-]", "_", out_name)
    out_path = WORK / out_name

    # ── Google Drive shortcut ──────────────────────────────────
    if "drive.google.com" in url:
        try:
            import gdown
            print("Google Drive link detected — downloading with gdown …")
            gdown.download(url, str(out_path), fuzzy=True)
        except Exception as exc:
            raise RuntimeError(f"gdown failed: {exc}") from exc
    else:
        # ── Standard HTTP download ─────────────────────────────
        print(f"Downloading to: {out_path}")
        headers = {
            "User-Agent": ("Mozilla/5.0 (X11; Linux x86_64) "
                           "AppleWebKit/537.36 Chrome/131 Safari/537.36"),
            "Accept": "*/*",
        }
        try:
            with requests.get(url, headers=headers, stream=True,
                              allow_redirects=True,
                              timeout=(30, 180)) as resp:
                print("HTTP         :", resp.status_code)
                print("Content-Type :", resp.headers.get("content-type"))
                print("Final URL    :", resp.url)
                resp.raise_for_status()

                # Try Content-Disposition if name is generic
                if out_name in ("download.bin", ""):
                    cd = resp.headers.get("content-disposition", "")
                    m = re.search(r'filename="?([^";\n]+)"?', cd)
                    if m:
                        out_name = re.sub(r"[^A-Za-z0-9._-]", "_",
                                          m.group(1).strip())
                        out_path = WORK / out_name

                tmp = out_path.with_suffix(out_path.suffix + ".part")
                with tmp.open("wb") as f:
                    for chunk in resp.iter_content(chunk_size=1024*1024):
                        if chunk:
                            f.write(chunk)
                tmp.replace(out_path)
        except requests.HTTPError as exc:
            out_path.unlink(missing_ok=True)
            raise RuntimeError(
                f"Download failed (HTTP {resp.status_code}). "
                "The link may have expired — paste a fresh one in Cell 2."
            ) from exc

    source_path = out_path

    if not source_path.exists() or source_path.stat().st_size < 1024:
        sz = source_path.stat().st_size if source_path.exists() else 0
        raise RuntimeError(f"File missing or too small ({sz} bytes).")
else:
    raise ValueError(f"Unknown INPUT_MODE: {INPUT_MODE!r}")

print(f"\nSource file : {source_path}")
print(f"Source size : {source_path.stat().st_size / 1e6:.1f} MB")

# ═══════════════════════════════════════════════════════════════
#  STEP 1 — Identify file type
# ═══════════════════════════════════════════════════════════════

def identify_filetype(filepath: Path) -> str:
    """Classify a file by its magic bytes / internal ZIP structure.

    Returns one of:
        apk | xapk | apks | apkm          — valid Android bundle
        zip_with_apk | zip_generic        — generic ZIP archive
        rar | 7z | gzip | bzip2 | xz     — non-ZIP archive
        dex | html | unknown              — other
    """
    head = filepath.read_bytes()[:32]
    hex16 = binascii.hexlify(head[:16]).decode()

    # ── Non-ZIP archive formats ────────────────────────────────
    if head[:6] == b'Rar!\x1a\x07':
        return 'rar'
    if head[:6] == b'\x37\x7a\xbc\xaf\x27\x1c':
        return '7z'
    if head[:2] == b'\x1f\x8b':
        return 'gzip'
    if head[:3] == b'BZh':
        return 'bzip2'
    if head[:6] == b'\xfd\x37\x7a\x58\x5a\x00':
        return 'xz'

    # ── ZIP container (could be APK, XAPK, APKS, or plain ZIP) ─
    if head[:4] in (b'PK\x03\x04', b'PK\x05\x06'):
        try:
            with zipfile.ZipFile(filepath) as z:
                names       = z.namelist()
                lower_names = [n.lower() for n in names]

                has_android_manifest = 'androidmanifest.xml' in lower_names
                has_manifest_json   = any(n.endswith('manifest.json')
                                          for n in lower_names)
                has_toc_pb          = any(n.endswith('toc.pb')
                                          for n in lower_names)
                has_info_json       = any(n.endswith('info.json')
                                          for n in lower_names)

                # Bundle indicators take priority
                if has_toc_pb:
                    return 'apks'
                if has_manifest_json:
                    return 'xapk'
                if has_info_json:
                    return 'apkm'
                # AndroidManifest.xml at root → this IS the APK
                if has_android_manifest:
                    return 'apk'

                # Check whether the ZIP *contains* APK/bundle files
                apk_inside = [n for n in names
                              if n.lower().endswith(('.apk', '.xapk', '.apks'))
                              and not n.endswith('/')]
                if apk_inside:
                    return 'zip_with_apk'

                return 'zip_generic'
        except zipfile.BadZipFile:
            return 'zip_corrupt'

    # ── DEX ────────────────────────────────────────────────────
    if head[:4] == b'dex\n':
        return 'dex'

    # ── HTML (expired / blocked link) ──────────────────────────
    if b'<html' in head.lower() or b'<!doctype' in head.lower():
        return 'html'

    return 'unknown'


filetype = identify_filetype(source_path)
print(f"Magic bytes : {binascii.hexlify(source_path.read_bytes()[:16]).decode()}")
print(f"Identified  : {filetype}")

if filetype == 'html':
    source_path.unlink(missing_ok=True)
    raise RuntimeError(
        "Server returned HTML — the link is probably expired or blocked."
    )

# ═══════════════════════════════════════════════════════════════
#  STEP 2 — Decompress archives (zip / rar / 7z / tar.*) and
#           find the actual APK / bundle inside
# ═══════════════════════════════════════════════════════════════

EXTRACT_DIR = WORK / "extracted"

def extract_archive(filepath, dest_dir, hint=None):
    """Extract *filepath* into *dest_dir*.

    Returns a sorted list of extracted file paths.
    """
    dest_dir = Path(dest_dir)
    dest_dir.mkdir(parents=True, exist_ok=True)
    ft = hint or identify_filetype(filepath)

    if ft in ('apk', 'xapk', 'apks', 'apkm', 'dex'):
        return [filepath]          # already a valid input

    if ft in ('zip_with_apk', 'zip_generic'):
        print(f"  Extracting ZIP  → {dest_dir}")
        with zipfile.ZipFile(filepath) as z:
            z.extractall(dest_dir)
        return sorted(p for p in dest_dir.rglob('*') if p.is_file())

    if ft == 'rar':
        print(f"  Extracting RAR  → {dest_dir}")
        try:
            import rarfile
            rarfile.UNRAR_TOOL = "unrar"
            with rarfile.RarFile(filepath) as rf:
                rf.extractall(dest_dir)
        except Exception:
            subprocess.run(
                ['unrar', 'x', '-o+', str(filepath),
                 str(dest_dir) + '/'],
                check=True
            )
        return sorted(p for p in dest_dir.rglob('*') if p.is_file())

    if ft == '7z':
        print(f"  Extracting 7z   → {dest_dir}")
        subprocess.run(
            ['7z', 'x', str(filepath), f'-o{dest_dir}', '-y'],
            check=True
        )
        return sorted(p for p in dest_dir.rglob('*') if p.is_file())

    if ft in ('gzip', 'bzip2', 'xz'):
        print(f"  Extracting tar  → {dest_dir}")
        try:
            with tarfile.open(filepath) as t:
                t.extractall(dest_dir)
            return sorted(p for p in dest_dir.rglob('*') if p.is_file())
        except tarfile.TarError:
            # Single compressed file, not a tar
            import gzip as _gz
            out_name = filepath.name
            for ext in ('.gz', '.bz2', '.xz'):
                out_name = out_name.replace(ext, '')
            if out_name == filepath.name:
                out_name = 'decompressed'
            out = dest_dir / out_name
            with _gz.open(filepath, 'rb') as fin, out.open('wb') as fout:
                shutil.copyfileobj(fin, fout)
            return [out]

    # Unknown — pass through
    print(f"  ⚠ Unknown type '{ft}' — passing through as-is.")
    return [filepath]


def find_apk_bundles(directory):
    """Return list of (path, filetype) for every APK / bundle
    found under *directory*, sorted largest-first."""
    results = []
    for p in directory.rglob('*'):
        if not p.is_file():
            continue
        ft = identify_filetype(p)
        if ft in ('apk', 'xapk', 'apks', 'apkm'):
            results.append((p, ft))
    results.sort(key=lambda x: x[0].stat().st_size, reverse=True)
    return results


# ── Clean previous extraction ──────────────────────────────────
if EXTRACT_DIR.exists():
    shutil.rmtree(EXTRACT_DIR)

# ── Already a valid APK / bundle? ──────────────────────────────
VALID_TYPES = ('apk', 'xapk', 'apks', 'apkm')

if filetype in VALID_TYPES:
    input_apk_path = source_path
    input_filetype = filetype
    print(f"\n✅ Input is already a {filetype} — no extraction needed.")

else:
    # ── Need to decompress ─────────────────────────────────────
    print(f"\n📦 File is a '{filetype}' archive — extracting …")
    extracted = extract_archive(source_path, EXTRACT_DIR, filetype)
    print(f"   Extracted {len(extracted)} file(s).")

    # ── Look for APK / bundle files ────────────────────────────
    apk_bundles = find_apk_bundles(EXTRACT_DIR)

    if apk_bundles:
        print(f"\n✅ Found {len(apk_bundles)} APK/bundle file(s):")
        for p, ft in apk_bundles:
            try:
                rel = p.relative_to(EXTRACT_DIR)
            except ValueError:
                rel = p
            print(f"   [{ft:>4}] {rel}  ({p.stat().st_size/1e6:.1f} MB)")

        input_apk_path, input_filetype = apk_bundles[0]
        print(f"\n   ➤ Using: {input_apk_path.name}  (type: {input_filetype})")

    else:
        # ── No APK found — check for nested archives ──────────
        print("\n🔍 No APK found directly. Checking for nested archives …")
        nested = []
        for p in extracted:
            ft = identify_filetype(p)
            if ft in ('zip_with_apk', 'zip_generic', 'rar', '7z',
                      'gzip', 'bzip2', 'xz'):
                nested.append((p, ft))

        if nested:
            print(f"   Found {len(nested)} nested archive(s) — "
                  "extracting recursively …")
            for i, (arch, ft) in enumerate(nested):
                ndest = EXTRACT_DIR / f"nested_{i}_{arch.stem}"
                extract_archive(arch, ndest, ft)

            # Re-scan everything
            apk_bundles = find_apk_bundles(EXTRACT_DIR)
            if apk_bundles:
                print(f"\n✅ Found {len(apk_bundles)} APK/bundle(s) "
                      "after recursive extraction:")
                for p, ft in apk_bundles:
                    print(f"   [{ft:>4}] {p}  ({p.stat().st_size/1e6:.1f} MB)")
                input_apk_path, input_filetype = apk_bundles[0]
                print(f"\n   ➤ Using: {input_apk_path.name}  "
                      f"(type: {input_filetype})")
            else:
                _raise_no_apk()
        else:
            # ── Last resort: match by extension ────────────────
            by_ext = sorted(
                (p for p in EXTRACT_DIR.rglob('*')
                 if p.is_file()
                 and p.suffix.lower() in ('.apk', '.xapk', '.apks')),
                key=lambda p: p.stat().st_size,
                reverse=True,
            )
            if by_ext:
                input_apk_path = by_ext[0]
                input_filetype = identify_filetype(input_apk_path)
                print(f"\n   ➤ Using (by extension): "
                      f"{input_apk_path.name}  ({input_filetype})")
            else:
                _raise_no_apk()


def _raise_no_apk():
    raise RuntimeError(
        "No APK / bundle files found in the extracted content.\n"
        "The archive may not contain an Android app, or the format "
        "is unsupported."
    )


# ── Summary ────────────────────────────────────────────────────
print(f"\n{'='*60}")
print(f"  INPUT APK/BUNDLE : {input_apk_path}")
print(f"  TYPE             : {input_filetype}")
print(f"  SIZE             : {input_apk_path.stat().st_size/1e6:.1f} MB")
print(f"{'='*60}")
````

````python
# -*- coding: utf-8 -*-
"""Cell 4: Prepare a single APK for decoding.

Reads `input_apk_path` set by Cell 3 (handles decompressed files).
If the input is a split-APK bundle, merges with APKEditor.
"""

import shutil, subprocess, zipfile
from pathlib import Path

WORK       = Path("/content/work")
merged_apk = WORK / "merged.apk"

if merged_apk.exists():
    merged_apk.unlink()

# ── Use the path identified by Cell 3 ──────────────────────────
if 'input_apk_path' in globals():
    out_path = Path(input_apk_path)
else:
    # Fallback: search WORK for a likely input file
    candidates = [
        p for p in WORK.iterdir()
        if p.is_file()
        and p.name not in {"merged.apk", "subscan_results.zip"}
        and not p.name.endswith(".part")
    ]
    if not candidates:
        raise FileNotFoundError(
            "No input file found. Run Cell 3 first."
        )
    out_path = max(candidates, key=lambda p: p.stat().st_mtime)

if not out_path.exists():
    raise FileNotFoundError(f"Input file not found: {out_path}")

print(f"Input: {out_path}")
print(f"Input size: {out_path.stat().st_size / 1e6:.1f} MB")

# ── Verify it is a ZIP container ───────────────────────────────
if not zipfile.is_zipfile(out_path):
    raise RuntimeError(
        f"Input is not a ZIP/APK: {out_path}\n"
        "If it was inside a compressed archive, check that Cell 3 "
        "extracted it correctly."
    )

with zipfile.ZipFile(out_path) as z:
    names        = z.namelist()
    lower_names  = [n.lower() for n in names]
    apk_members  = [n for n in names
                    if n.lower().endswith('.apk') and not n.endswith('/')]

    has_manifest = any(n.endswith('manifest.json') for n in lower_names)
    has_toc      = any(n.endswith('toc.pb')       for n in lower_names)
    has_info     = any(n.endswith('info.json')     for n in lower_names)

print(f"APKs inside bundle  : {len(apk_members)}")
print(f"manifest.json (XAPK): {has_manifest}")
print(f"info.json   (APKM)  : {has_info}")
print(f"toc.pb      (APKS)  : {has_toc}")

# ── Normal APK ──────────────────────────────────────────────────
if "AndroidManifest.xml" in names and not apk_members:
    shutil.copy2(out_path, merged_apk)
    print("Input is a normal APK.")

elif "AndroidManifest.xml" in names and len(apk_members) == 0:
    shutil.copy2(out_path, merged_apk)
    print("Input is a normal APK.")

# ── ZIP wrapper with one APK ───────────────────────────────────
elif len(apk_members) == 1 and not (has_manifest or has_info or has_toc):
    with zipfile.ZipFile(out_path) as z:
        z.extract(apk_members[0], WORK)
    extracted = WORK / apk_members[0]
    shutil.copy2(extracted, merged_apk)
    extracted.unlink(missing_ok=True)
    print("Extracted the single APK from the ZIP wrapper.")

# ── Split-APK / XAPK / APKS / APKM bundle ─────────────────────
else:
    print("Bundle detected — merging with APKEditor …")
    cmd = [
        "java", "-jar", "/content/APKEditor.jar",
        "m", "-i", str(out_path), "-o", str(merged_apk), "-f"
    ]
    r = subprocess.run(cmd, capture_output=True, text=True)

    if r.stdout:
        print(r.stdout[-4000:])
    if r.returncode != 0:
        print("STDERR:")
        print(r.stderr[-4000:])
        raise RuntimeError(
            "APKEditor could not merge the bundle. "
            "Check the bundle format and APKEditor output above."
        )

if not merged_apk.exists() or merged_apk.stat().st_size < 10_000:
    raise RuntimeError("merged.apk was not created correctly.")

print(
    f"\n✅ Merged APK: {merged_apk}  "
    f"({merged_apk.stat().st_size / 1e6:.1f} MB)"
)
````

````python
# CELL 5 — Decode with apk.sh
import os
import shutil
import subprocess
from pathlib import Path

WORK = Path("/content/work")
merged_apk = WORK / "merged.apk"
decoded_dir = WORK / "merged"

if not merged_apk.exists():
    raise FileNotFoundError("merged.apk not found. Run Cell 4 first.")

if decoded_dir.exists():
    shutil.rmtree(decoded_dir)

os.chdir(WORK)

print("Decoding with apk.sh...")
r = subprocess.run(
    [
        "bash", "/content/apk.sh/apk.sh",
        "decode", str(merged_apk)
    ],
    capture_output=True,
    text=True
)

if r.stdout:
    print(r.stdout[-5000:])
if r.returncode != 0:
    print("STDERR:")
    print(r.stderr[-5000:])
    raise RuntimeError("apk.sh decode failed.")

# apk.sh normally creates /content/work/merged.
if not decoded_dir.exists():
    # Fallback: locate a recently created directory containing smali.
    cands = [
        d for d in WORK.iterdir()
        if d.is_dir() and any(d.glob("smali*"))
    ]
    if not cands:
        raise FileNotFoundError(
            "Decoded directory not found. Inspect apk.sh output above."
        )
    decoded_dir = max(cands, key=lambda d: d.stat().st_mtime)

smali_dirs = [
    p for p in decoded_dir.glob("smali*")
    if p.is_dir()
]

print("Decoded dir:", decoded_dir)
print("Smali dirs:", [str(p.relative_to(decoded_dir)) for p in smali_dirs])

if not smali_dirs:
    raise RuntimeError(
        "No smali directories were produced. "
        "The APK may be obfuscated or apk.sh/apktool decoding failed."
    )
````

````python
# CELL 6 — Analyze subscription-related code
import re
import json
from collections import defaultdict
from pathlib import Path

decoded_dir = Path(decoded_dir)

BILLING_SDK_PATTERNS = {
    "Google Play Billing": [
        r"Lcom/android/billingclient/api/",
        r"\bBillingClient\b",
        r"\bqueryPurchasesAsync\b",
        r"\bqueryPurchases\b",
        r"\blaunchBillingFlow\b",
        r"ProductType\.SUBS",
        r"SkuType\.SUBS",
        r"\bgetPurchaseState\b",
        r"\bgetPurchaseToken\b",
        r"\bgetSubscriptionId\b",
        r"\backnowledgePurchase\b",
    ],
    "RevenueCat": [
        r"Lcom/revenuecat/purchases/",
        r"Purchases\.sharedInstance",
        r"\bCustomerInfo\b",
        r"\bEntitlementInfo\b",
        r"\bgetEntitlements\b",
    ],
    "Amazon IAP": [
        r"Lcom/amazon/device/iap/",
        r"\bPurchasingService\b",
        r"\bgetPurchaseUpdates\b",
    ],
}

PREMIUM_STATE_PATTERNS = [
    r"\bisVip\b", r"\bisVIP\b", r"\bisPremium\b", r"\bisPro\b",
    r"\bisSubscribed\b", r"\bisSubscriptionActive\b",
    r"\bhasSubscription\b", r"\bcheckSubscription\b",
    r"\bcheckVip\b", r"\bcheckPremium\b",
    r"\bverifyPurchase\b", r"\bverifySubscription\b",
    r"\bvalidatePurchase\b", r"\bgetVipType\b",
    r"\bgetSubscriptionType\b", r"\bgetPlanType\b",
    r"\bgetMembershipType\b", r"\bgetPremiumType\b",
    r"\bsubscriptionType\b", r"\bvipType\b",
    r"\bplanType\b", r"\bmembershipType\b",
]

SUBTYPE_PATTERNS = [
    r"\bSUBSCRIPTION_\w+\b", r"\bPERIOD_\w+\b",
    r"\bBILLING_\w+\b", r"\bSKU_\w+\b",
    r"\bINAPP_\w+\b", r"\bMONTHLY\b", r"\bYEARLY\b",
    r"\bWEEKLY\b", r"\bLIFETIME\b", r"\bFOREVER\b", r"\bTRIAL\b",
]

# Skip common third-party framework namespaces to reduce noise.
SKIP = (
    "android/", "androidx/", "com/google/",
    "kotlin/", "kotlinx/"
)

def iter_smali(root):
    for d in root.glob("smali*"):
        if not d.is_dir():
            continue
        for f in d.rglob("*.smali"):
            rel = str(f.relative_to(d)).replace("\\", "/")
            if rel.startswith(SKIP):
                continue
            yield f

def snippet(text, pos, max_lines=30):
    lines = text.splitlines()
    if not lines:
        return ""

    offset = 0
    line_no = 0
    for i, line in enumerate(lines):
        if offset + len(line) >= pos:
            line_no = i
            break
        offset += len(line) + 1

    start = line_no
    for i in range(line_no, -1, -1):
        if lines[i].strip().startswith(".method"):
            start = i
            break

    end = line_no
    for i in range(line_no, len(lines)):
        if lines[i].strip().startswith(".end method"):
            end = i
            break
    else:
        end = min(len(lines) - 1, line_no + max_lines)

    if end - start > max_lines:
        end = start + max_lines

    return "\n".join(lines[start:end + 1])

results = {
    "billing_sdk": defaultdict(list),
    "premium": [],
    "subtype": [],
}

files = list(iter_smali(decoded_dir))
print(f"Scanning {len(files)} smali files...")

for f in files:
    try:
        text = f.read_text(encoding="utf-8", errors="replace")
    except Exception:
        continue

    rel_file = str(f.relative_to(decoded_dir))

    for sdk, patterns in BILLING_SDK_PATTERNS.items():
        for pattern in patterns:
            for m in re.finditer(pattern, text, re.IGNORECASE):
                results["billing_sdk"][sdk].append({
                    "file": rel_file,
                    "match": m.group(0),
                    "snippet": snippet(text, m.start()),
                })

    for pattern in PREMIUM_STATE_PATTERNS:
        for m in re.finditer(pattern, text, re.IGNORECASE):
            results["premium"].append({
                "file": rel_file,
                "match": m.group(0),
                "snippet": snippet(text, m.start()),
            })

    for pattern in SUBTYPE_PATTERNS:
        for m in re.finditer(pattern, text, re.IGNORECASE):
            results["subtype"].append({
                "file": rel_file,
                "match": m.group(0),
                "snippet": snippet(text, m.start()),
            })

print("\n" + "=" * 70)
print("SUBSCRIPTION ANALYSIS REPORT")
print("=" * 70)

print("\n[1] Billing SDK")
if not results["billing_sdk"]:
    print("  None detected.")
else:
    for sdk, hits in results["billing_sdk"].items():
        print(f"  {sdk}: {len(hits)} hits")

print("\n[2] Premium / Subscription State Methods")
if not results["premium"]:
    print("  None found.")
else:
    grouped = defaultdict(list)
    for h in results["premium"]:
        grouped[h["match"].lower()].append(h)
    for name in sorted(grouped):
        hits = grouped[name]
        print(
            f"  {name} ({len(hits)}) "
            f"e.g. {hits[0]['file']}"
        )

print("\n[3] Subscription-Type Constants")
if not results["subtype"]:
    print("  None found.")
else:
    grouped = defaultdict(list)
    for h in results["subtype"]:
        grouped[h["match"].upper()].append(h)
    for name in sorted(grouped):
        hits = grouped[name]
        print(
            f"  {name} ({len(hits)}) "
            f"e.g. {hits[0]['file']}"
        )

print("\n[4] Conclusion")
if results["billing_sdk"]:
    best = max(
        results["billing_sdk"].items(),
        key=lambda kv: len(kv[1])
    )
    print(f"  Billing SDK detected: {best[0]}")
    print("  Subscription-related code was found in the decoded app.")
elif results["premium"]:
    print("  No known billing SDK detected.")
    print("  Premium/subscription state code exists; it may be custom or server-side.")
else:
    print("  No clear subscription mechanism found in the scanned Smali.")
    print("  Native libraries, obfuscated code, or DEX analysis may be required.")
````

````python
# CELL 10 — Refined scan targeting the app's own subscription code
# The Google Play Billing classes are just the SDK. The app's own logic
# lives in com/celsys/*. Let's find it.

import re
from collections import defaultdict
from pathlib import Path

decoded_dir = Path(decoded_dir)

# Classes worth inspecting
INTERESTING_PREFIXES = [
    "com/celsys/",
    "com/celsys/velegacyandroidhelpers/",
]

# Method-signature patterns for subscription checks
METHOD_PATTERNS = [
    r"\.method[^\n]*\b(is[A-Z]\w*|has[A-Z]\w*|get[A-Z]\w*|check\w*|verify\w*|validate\w*|can\w*)\s*\(",
]

purchase_classes = []
for d in decoded_dir.glob("smali*"):
    if not d.is_dir():
        continue
    for f in d.rglob("*.smali"):
        rel = str(f.relative_to(decoded_dir)).replace("\\", "/")
        if any(rel.startswith(p) or p in rel for p in INTERESTING_PREFIXES):
            purchase_classes.append(f)

print(f"Found {len(purchase_classes)} classes under com/celsys/")
print()

# For each, list methods and fields
for f in purchase_classes:
    text = f.read_text(encoding="utf-8", errors="replace")
    if "purchase" not in text.lower() and "subscription" not in text.lower() and "billing" not in text.lower():
        continue
    rel = str(f.relative_to(decoded_dir))
    methods = re.findall(r"^\.method\s+(?:public\s+|private\s+|protected\s+|static\s+|final\s+|abstract\s+|synthetic\s+)*(\w+)\s*\(([^)]*)\)([^\n]*)", text, re.MULTILINE)
    fields = re.findall(r"^\.field\s+(?:public\s+|private\s+|protected\s+|static\s+|final\s+)*(\w+):([^\s]+)", text, re.MULTILINE)

    if methods or fields:
        print(f"── {rel}")
        if fields:
            print(f"   Fields: {', '.join(f'{n}:{t}' for n,t in fields[:10])}")
        if methods:
            print(f"   Methods ({len(methods)}):")
            for name, args, ret in methods[:20]:
                ret = ret.strip()
                print(f"      {name}({args}){ret}")
        print()
````
