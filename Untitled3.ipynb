```python
# -*- coding: utf-8 -*-
"""Cell 1: Environment setup — Java, apktool, apk.sh, APKEditor, archive tools."""

import os, sys, subprocess, json, urllib.request
from pathlib import Path

def run_cmd(cmd, check=True, **kwargs):
    print("$", " ".join(map(str, cmd)))
    return subprocess.run(cmd, check=check, **kwargs)

# ── System packages ──────────────────────────────────────────────
run_cmd(["apt-get", "update", "-qq"])
run_cmd([
    "apt-get", "install", "-y", "-qq",
    "openjdk-17-jre-headless",
    "unzip",          # ZIP extraction
    "unrar",          # RAR extraction
    "p7zip-full",     # 7z extraction
    "curl", "git",
], stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL)

# Python packages for archive handling & Google Drive
run_cmd([sys.executable, "-m", "pip", "install", "-q",
         "rarfile", "gdown", "requests"], check=False)

print("Java:")
run_cmd(["java", "-version"], check=False)

# ── apktool 2.9.3 ───────────────────────────────────────────────
APKTOOL_JAR = Path("/usr/local/bin/apktool.jar")
APKTOOL_BIN = Path("/usr/local/bin/apktool")

if not APKTOOL_JAR.exists() or APKTOOL_JAR.stat().st_size < 100_000:
    run_cmd([
        "curl", "-fL", "--retry", "3",
        "-o", str(APKTOOL_JAR),
        "https://github.com/iBotPeaches/Apktool/releases/download/"
        "v2.9.3/apktool_2.9.3.jar"
    ])

APKTOOL_BIN.write_text(
    '#!/bin/bash\nexec java -jar /usr/local/bin/apktool.jar "$@"\n'
)
APKTOOL_BIN.chmod(0o755)

r = subprocess.run(["apktool", "--version"],
                   capture_output=True, text=True)
print("apktool:", r.stdout.strip() or r.stderr.strip())

# ── apk.sh ──────────────────────────────────────────────────────
APKSH_DIR = Path("/content/apk.sh")
if not APKSH_DIR.exists():
    run_cmd(["git", "clone", "--depth=1",
             "https://github.com/ax/apk.sh.git", str(APKSH_DIR)])
(APKSH_DIR / "apk.sh").chmod(0o755)
print("apk.sh:", APKSH_DIR / "apk.sh")

# ── APKEditor (latest release .jar) ────────────────────────────
APKEDITOR = Path("/content/APKEditor.jar")

def download_apkeditor():
    api = ("https://api.github.com/repos/"
           "REAndroid/APKEditor/releases/latest")
    req = urllib.request.Request(api,
                                headers={"User-Agent": "Mozilla/5.0"})
    with urllib.request.urlopen(req, timeout=30) as resp:
        release = json.load(resp)

    jar_assets = [a for a in release.get("assets", [])
                  if a.get("name", "").lower().endswith(".jar")]
    if not jar_assets:
        raise RuntimeError("No .jar asset found on APKEditor releases.")

    asset   = jar_assets[0]
    url     = asset["browser_download_url"]
    print("APKEditor release :", release.get("tag_name"))
    print("APKEditor asset  :", asset["name"])

    req = urllib.request.Request(url,
                                headers={"User-Agent": "Mozilla/5.0"})
    with urllib.request.urlopen(req, timeout=120) as resp, \
         APKEDITOR.open("wb") as f:
        while True:
            chunk = resp.read(1024 * 1024)
            if not chunk:
                break
            f.write(chunk)

if not APKEDITOR.exists() or APKEDITOR.stat().st_size < 100_000:
    if APKEDITOR.exists():
        APKEDITOR.unlink()
    download_apkeditor()

print("APKEditor:", APKEDITOR,
      f"({APKEDITOR.stat().st_size / 1e6:.1f} MB)")
print("\n✅ Setup complete.")
```

```python
# -*- coding: utf-8 -*-
"""Cell 3: Download (or receive local/upload) for Link 1 and Link 2, decompress archives,
identify the real APK/bundle inside, and set entries_info."""

import os, sys, re, shutil, binascii, zipfile, subprocess, tarfile
from pathlib import Path
from urllib.parse import urlparse

try:
    import requests
except ImportError:
    subprocess.run([sys.executable, "-m", "pip", "install", "-q", "requests"], check=False)
    import requests

BASE_WORK = Path("/content/work")
BASE_WORK.mkdir(parents=True, exist_ok=True)
os.chdir(BASE_WORK)

# Construct entry list from LINK_1 and LINK_2
ENTRY_URLS = [u.strip() for u in [globals().get('LINK_1', ''), globals().get('LINK_2', '')] if u.strip()]

entries_info = []
targets = []

if INPUT_MODE == "url":
    targets = [("url", u) for u in ENTRY_URLS]
elif INPUT_MODE == "local":
    paths = [p.strip() for p in globals().get('LOCAL_PATHS', '').strip().splitlines() if p.strip()]
    targets = [("local", p) for p in paths]
elif INPUT_MODE == "upload":
    try:
        from google.colab import files as colab_files
    except ImportError:
        raise RuntimeError("Upload mode requires Google Colab.")
    print("Opening file picker …")
    uploaded = colab_files.upload()
    if not uploaded:
        raise RuntimeError("No file was uploaded.")
    targets = [("upload", (fname, data)) for fname, data in uploaded.items()]
else:
    raise ValueError(f"Unknown INPUT_MODE: {INPUT_MODE!r}")

print(f"Processing {len(targets)} entry target(s) ...\n")

def identify_filetype(filepath: Path) -> str:
    head = filepath.read_bytes()[:32]
    if head[:6] == b'Rar!\x1a\x07':
        return 'rar'
    if head[:6] == b'\x37\x7a\xbc\xaf\x27\x1c':
        return '7z'
    if head[:2] == b'\x1f\x8b':
        return 'gzip'
    if head[:3] == b'BZh':
        return 'bzip2'
    if head[:6] == b'\xfd\x37\x7a\x58\x5a\x00':
        return 'xz'

    if head[:4] in (b'PK\x03\x04', b'PK\x05\x06'):
        try:
            with zipfile.ZipFile(filepath) as z:
                names = z.namelist()
                lower_names = [n.lower() for n in names]
                has_android_manifest = 'androidmanifest.xml' in lower_names
                has_manifest_json = any(n.endswith('manifest.json') for n in lower_names)
                has_toc_pb = any(n.endswith('toc.pb') for n in lower_names)
                has_info_json = any(n.endswith('info.json') for n in lower_names)

                if has_toc_pb:
                    return 'apks'
                if has_manifest_json:
                    return 'xapk'
                if has_info_json:
                    return 'apkm'
                if has_android_manifest:
                    return 'apk'

                apk_inside = [n for n in names if n.lower().endswith(('.apk', '.xapk', '.apks')) and not n.endswith('/')]
                if apk_inside:
                    return 'zip_with_apk'
                return 'zip_generic'
        except zipfile.BadZipFile:
            return 'zip_corrupt'

    if head[:4] == b'dex\n':
        return 'dex'
    if b'<html' in head.lower() or b'<!doctype' in head.lower():
        return 'html'
    return 'unknown'

def extract_archive(filepath, dest_dir, hint=None):
    dest_dir = Path(dest_dir)
    dest_dir.mkdir(parents=True, exist_ok=True)
    ft = hint or identify_filetype(filepath)

    if ft in ('apk', 'xapk', 'apks', 'apkm', 'dex'):
        return [filepath]

    if ft in ('zip_with_apk', 'zip_generic'):
        print(f"  Extracting ZIP  → {dest_dir}")
        with zipfile.ZipFile(filepath) as z:
            z.extractall(dest_dir)
        return sorted(p for p in dest_dir.rglob('*') if p.is_file())

    if ft == 'rar':
        print(f"  Extracting RAR  → {dest_dir}")
        try:
            import rarfile
            rarfile.UNRAR_TOOL = "unrar"
            with rarfile.RarFile(filepath) as rf:
                rf.extractall(dest_dir)
        except Exception:
            subprocess.run(['unrar', 'x', '-o+', str(filepath), str(dest_dir) + '/'], check=True)
        return sorted(p for p in dest_dir.rglob('*') if p.is_file())

    if ft == '7z':
        print(f"  Extracting 7z   → {dest_dir}")
        subprocess.run(['7z', 'x', str(filepath), f'-o{dest_dir}', '-y'], check=True)
        return sorted(p for p in dest_dir.rglob('*') if p.is_file())

    if ft in ('gzip', 'bzip2', 'xz'):
        print(f"  Extracting tar  → {dest_dir}")
        try:
            with tarfile.open(filepath) as t:
                t.extractall(dest_dir)
            return sorted(p for p in dest_dir.rglob('*') if p.is_file())
        except tarfile.TarError:
            import gzip as _gz
            out_name = filepath.name
            for ext in ('.gz', '.bz2', '.xz'):
                out_name = out_name.replace(ext, '')
            if out_name == filepath.name:
                out_name = 'decompressed'
            out = dest_dir / out_name
            with _gz.open(filepath, 'rb') as fin, out.open('wb') as fout:
                shutil.copyfileobj(fin, fout)
            return [out]

    print(f"  ⚠ Unknown type '{ft}' — passing through as-is.")
    return [filepath]

def find_apk_bundles(directory):
    results = []
    for p in directory.rglob('*'):
        if not p.is_file():
            continue
        ft = identify_filetype(p)
        if ft in ('apk', 'xapk', 'apks', 'apkm'):
            results.append((p, ft))
    results.sort(key=lambda x: x[0].stat().st_size, reverse=True)
    return results

for idx, (mode, payload) in enumerate(targets, 1):
    entry_dir = BASE_WORK / f"entry_{idx}"
    if entry_dir.exists():
        shutil.rmtree(entry_dir)
    entry_dir.mkdir(parents=True, exist_ok=True)
    os.chdir(entry_dir)

    print(f"=== ENTRY {idx} (Link {idx}) ===")
    if mode == "url":
        url = payload.strip()
        out_name = Path(urlparse(url).path).name or f"download_{idx}.bin"
        out_name = re.sub(r"[^A-Za-z0-9._-]", "_", out_name)
        out_path = entry_dir / out_name

        if "drive.google.com" in url:
            try:
                import gdown
                print("Google Drive link detected — downloading with gdown …")
                gdown.download(url, str(out_path), fuzzy=True)
            except Exception as exc:
                raise RuntimeError(f"gdown failed: {exc}") from exc
        else:
            print(f"Downloading to: {out_path}")
            headers = {
                "User-Agent": ("Mozilla/5.0 (X11; Linux x86_64) AppleWebKit/537.36 Chrome/131 Safari/537.36"),
                "Accept": "*/*",
            }
            try:
                with requests.get(url, headers=headers, stream=True, allow_redirects=True, timeout=(30, 180)) as resp:
                    print("HTTP         :", resp.status_code)
                    print("Content-Type :", resp.headers.get("content-type"))
                    resp.raise_for_status()

                    cd = resp.headers.get("content-disposition", "")
                    m = re.search(r'filename="?([^";\n]+)"?', cd)
                    if m:
                        out_name = re.sub(r"[^A-Za-z0-9._-]", "_", m.group(1).strip())
                        out_path = entry_dir / out_name

                    tmp = out_path.with_suffix(out_path.suffix + ".part")
                    with tmp.open("wb") as f:
                        for chunk in resp.iter_content(chunk_size=1024*1024):
                            if chunk:
                                f.write(chunk)
                    tmp.replace(out_path)
            except requests.HTTPError as exc:
                out_path.unlink(missing_ok=True)
                raise RuntimeError(f"Download failed for Link {idx} (HTTP {resp.status_code}).") from exc

        source_path = out_path

    elif mode == "local":
        source_path = Path(payload.strip())
        if not source_path.exists():
            raise FileNotFoundError(f"Local file not found: {source_path}")
        dest = entry_dir / source_path.name
        shutil.copy2(source_path, dest)
        source_path = dest

    elif mode == "upload":
        fname, data = payload
        dest = entry_dir / re.sub(r"[^A-Za-z0-9._-]", "_", fname)
        dest.write_bytes(data)
        source_path = dest

    filetype = identify_filetype(source_path)
    print(f"Source file : {source_path.name}")
    print(f"Source size : {source_path.stat().st_size / 1e6:.1f} MB")
    print(f"Identified  : {filetype}")

    if filetype == 'html':
        raise RuntimeError(f"Link {idx} returned HTML content — link may be expired.")

    EXTRACT_DIR = entry_dir / "extracted"
    VALID_TYPES = ('apk', 'xapk', 'apks', 'apkm')

    if filetype in VALID_TYPES:
        input_apk_path = source_path
        input_filetype = filetype
    else:
        extracted = extract_archive(source_path, EXTRACT_DIR, filetype)
        apk_bundles = find_apk_bundles(EXTRACT_DIR)
        if apk_bundles:
            input_apk_path, input_filetype = apk_bundles[0]
        else:
            nested = []
            for p in extracted:
                ft = identify_filetype(p)
                if ft in ('zip_with_apk', 'zip_generic', 'rar', '7z', 'gzip', 'bzip2', 'xz'):
                    nested.append((p, ft))
            if nested:
                for n_idx, (arch, ft) in enumerate(nested):
                    ndest = EXTRACT_DIR / f"nested_{n_idx}_{arch.stem}"
                    extract_archive(arch, ndest, ft)
                apk_bundles = find_apk_bundles(EXTRACT_DIR)
                if apk_bundles:
                    input_apk_path, input_filetype = apk_bundles[0]
                else:
                    raise RuntimeError(f"No APK found in Link {idx}.")
            else:
                by_ext = sorted([p for p in EXTRACT_DIR.rglob('*') if p.is_file() and p.suffix.lower() in ('.apk', '.xapk', '.apks')],
                                key=lambda p: p.stat().st_size, reverse=True)
                if by_ext:
                    input_apk_path = by_ext[0]
                    input_filetype = identify_filetype(input_apk_path)
                else:
                    raise RuntimeError(f"No APK found in Link {idx}.")

    info = {
        "entry_idx": idx,
        "entry_dir": entry_dir,
        "source_path": source_path,
        "input_apk_path": input_apk_path,
        "input_filetype": input_filetype,
    }
    entries_info.append(info)
    print(f"✅ Link {idx} input ready: {input_apk_path.name} ({input_filetype})\n")

os.chdir(BASE_WORK)
```

```python
# -*- coding: utf-8 -*-
"""Cell 4: Prepare single APK for decoding for each entry."""

import shutil, subprocess, zipfile
from pathlib import Path

BASE_WORK = Path("/content/work")

for info in entries_info:
    idx = info["entry_idx"]
    entry_dir = info["entry_dir"]
    out_path = Path(info["input_apk_path"])
    merged_apk = entry_dir / "merged.apk"

    if merged_apk.exists():
        merged_apk.unlink()

    print(f"=== Merging/Preparing Entry {idx} ===")
    print(f"Input: {out_path.name} ({out_path.stat().st_size / 1e6:.1f} MB)")

    if not zipfile.is_zipfile(out_path):
        raise RuntimeError(f"Input for Entry {idx} is not a valid ZIP/APK: {out_path}")

    with zipfile.ZipFile(out_path) as z:
        names = z.namelist()
        lower_names = [n.lower() for n in names]
        apk_members = [n for n in names if n.lower().endswith('.apk') and not n.endswith('/')]
        has_manifest = any(n.endswith('manifest.json') for n in lower_names)
        has_toc = any(n.endswith('toc.pb') for n in lower_names)
        has_info = any(n.endswith('info.json') for n in lower_names)

    if "AndroidManifest.xml" in names and not apk_members:
        shutil.copy2(out_path, merged_apk)
        print(f"Entry {idx} is a normal APK.")
    elif len(apk_members) == 1 and not (has_manifest or has_info or has_toc):
        with zipfile.ZipFile(out_path) as z:
            z.extract(apk_members[0], entry_dir)
        extracted = entry_dir / apk_members[0]
        shutil.copy2(extracted, merged_apk)
        extracted.unlink(missing_ok=True)
        print(f"Extracted single APK for Entry {idx}.")
    else:
        print(f"Bundle detected for Entry {idx} — merging with APKEditor …")
        cmd = [
            "java", "-jar", "/content/APKEditor.jar",
            "m", "-i", str(out_path), "-o", str(merged_apk), "-f"
        ]
        r = subprocess.run(cmd, capture_output=True, text=True)
        if r.returncode != 0:
            print("STDERR:", r.stderr[-2000:])
            raise RuntimeError(f"APKEditor failed to merge Entry {idx}.")

    if not merged_apk.exists() or merged_apk.stat().st_size < 10_000:
        raise RuntimeError(f"merged.apk was not created correctly for Entry {idx}.")

    info["merged_apk"] = merged_apk
    print(f"✅ Entry {idx} merged APK: {merged_apk} ({merged_apk.stat().st_size / 1e6:.1f} MB)\n")
```

```python
# CELL 5 — Decode each entry with apk.sh
import os, shutil, subprocess
from pathlib import Path

BASE_WORK = Path("/content/work")

for info in entries_info:
    idx = info["entry_idx"]
    entry_dir = info["entry_dir"]
    merged_apk = info["merged_apk"]
    decoded_dir = entry_dir / "merged"

    if decoded_dir.exists():
        shutil.rmtree(decoded_dir)

    os.chdir(entry_dir)
    print(f"=== Decoding Entry {idx} with apk.sh ===")
    r = subprocess.run(
        ["bash", "/content/apk.sh/apk.sh", "decode", str(merged_apk)],
        capture_output=True, text=True
    )

    if r.returncode != 0:
        print("STDERR:", r.stderr[-2000:])
        raise RuntimeError(f"apk.sh decode failed for Entry {idx}.")

    if not decoded_dir.exists():
        cands = [d for d in entry_dir.iterdir() if d.is_dir() and any(d.glob("smali*"))]
        if not cands:
            raise FileNotFoundError(f"Decoded dir not found for Entry {idx}.")
        decoded_dir = max(cands, key=lambda d: d.stat().st_mtime)

    smali_dirs = [p for p in decoded_dir.glob("smali*") if p.is_dir()]
    info["decoded_dir"] = decoded_dir
    print(f"Decoded dir for Entry {idx}: {decoded_dir}")
    print(f"Smali dirs: {[str(p.relative_to(decoded_dir)) for p in smali_dirs]}\n")

os.chdir(BASE_WORK)
```

```python
# CELL 6 — Analyze subscription code & perform comparison across entries
import re, json, hashlib
from collections import defaultdict
from pathlib import Path

BILLING_SDK_PATTERNS = {
    "Google Play Billing": [
        r"Lcom/android/billingclient/api/",
        r"\bBillingClient\b",
        r"\bqueryPurchasesAsync\b",
        r"\bqueryPurchases\b",
        r"\blaunchBillingFlow\b",
        r"ProductType\.SUBS",
        r"SkuType\.SUBS",
        r"\bgetPurchaseState\b",
        r"\bgetPurchaseToken\b",
        r"\bgetSubscriptionId\b",
        r"\backnowledgePurchase\b",
    ],
    "RevenueCat": [
        r"Lcom/revenuecat/purchases/",
        r"Purchases\.sharedInstance",
        r"\bCustomerInfo\b",
        r"\bEntitlementInfo\b",
        r"\bgetEntitlements\b",
    ],
    "Amazon IAP": [
        r"Lcom/amazon/device/iap/",
        r"\bPurchasingService\b",
        r"\bgetPurchaseUpdates\b",
    ],
}

PREMIUM_STATE_PATTERNS = [
    r"\bisVip\b", r"\bisVIP\b", r"\bisPremium\b", r"\bisPro\b",
    r"\bisSubscribed\b", r"\bisSubscriptionActive\b",
    r"\bhasSubscription\b", r"\bcheckSubscription\b",
    r"\bcheckVip\b", r"\bcheckPremium\b",
    r"\bverifyPurchase\b", r"\bverifySubscription\b",
    r"\bvalidatePurchase\b", r"\bgetVipType\b",
    r"\bgetSubscriptionType\b", r"\bgetPlanType\b",
    r"\bgetMembershipType\b", r"\bgetPremiumType\b",
    r"\bsubscriptionType\b", r"\bvipType\b",
    r"\bplanType\b", r"\bmembershipType\b",
]

SUBTYPE_PATTERNS = [
    r"\bSUBSCRIPTION_\w+\b", r"\bPERIOD_\w+\b",
    r"\bBILLING_\w+\b", r"\bSKU_\w+\b",
    r"\bINAPP_\w+\b", r"\bMONTHLY\b", r"\bYEARLY\b",
    r"\bWEEKLY\b", r"\bLIFETIME\b", r"\bFOREVER\b", r"\bTRIAL\b",
]

SKIP = ("android/", "androidx/", "com/google/", "kotlin/", "kotlinx/")

def iter_smali(root):
    for d in root.glob("smali*"):
        if not d.is_dir():
            continue
        for f in d.rglob("*.smali"):
            rel = str(f.relative_to(d)).replace("\\", "/")
            if rel.startswith(SKIP):
                continue
            yield f

def snippet(text, pos, max_lines=30):
    lines = text.splitlines()
    if not lines:
        return ""
    offset = 0
    line_no = 0
    for i, line in enumerate(lines):
        if offset + len(line) >= pos:
            line_no = i
            break
        offset += len(line) + 1
    start = line_no
    for i in range(line_no, -1, -1):
        if lines[i].strip().startswith(".method"):
            start = i
            break
    end = line_no
    for i in range(line_no, len(lines)):
        if lines[i].strip().startswith(".end method"):
            end = i
            break
    else:
        end = min(len(lines) - 1, line_no + max_lines)
    if end - start > max_lines:
        end = start + max_lines
    return "\n".join(lines[start:end + 1])

def get_file_sha256(filepath):
    h = hashlib.sha256()
    with open(filepath, "rb") as f:
        while chunk := f.read(8192):
            h.update(chunk)
    return h.hexdigest()

analysis_results = {}

for info in entries_info:
    idx = info["entry_idx"]
    decoded_dir = Path(info["decoded_dir"])
    merged_apk = Path(info["merged_apk"])

    res = {
        "billing_sdk": defaultdict(list),
        "premium": [],
        "subtype": [],
        "apk_size": merged_apk.stat().st_size,
        "apk_sha256": get_file_sha256(merged_apk),
        "file_tree": {},
    }

    for f in decoded_dir.rglob("*"):
        if f.is_file():
            rel = str(f.relative_to(decoded_dir)).replace("\\", "/")
            res["file_tree"][rel] = {
                "size": f.stat().st_size,
                "sha256": get_file_sha256(f) if f.stat().st_size < 10_000_000 else "large_file"
            }

    files = list(iter_smali(decoded_dir))
    print(f"Entry {idx}: Scanning {len(files)} smali files...")

    for f in files:
        try:
            text = f.read_text(encoding="utf-8", errors="replace")
        except Exception:
            continue
        rel_file = str(f.relative_to(decoded_dir))

        for sdk, patterns in BILLING_SDK_PATTERNS.items():
            for pattern in patterns:
                for m in re.finditer(pattern, text, re.IGNORECASE):
                    res["billing_sdk"][sdk].append({
                        "file": rel_file,
                        "match": m.group(0),
                        "snippet": snippet(text, m.start()),
                    })

        for pattern in PREMIUM_STATE_PATTERNS:
            for m in re.finditer(pattern, text, re.IGNORECASE):
                res["premium"].append({
                    "file": rel_file,
                    "match": m.group(0),
                    "snippet": snippet(text, m.start()),
                })

        for pattern in SUBTYPE_PATTERNS:
            for m in re.finditer(pattern, text, re.IGNORECASE):
                res["subtype"].append({
                    "file": rel_file,
                    "match": m.group(0),
                    "snippet": snippet(text, m.start()),
                })

    analysis_results[idx] = res

print("\n✅ Subscription and file tree analysis completed for all entries.")
```

```python
# CELL 10 — Refined scan targeting the app's own subscription code
# The Google Play Billing classes are just the SDK. The app's own logic
# lives in com/celsys/*. Let's find it.

import re
from collections import defaultdict
from pathlib import Path

decoded_dir = Path(decoded_dir)

# Classes worth inspecting
INTERESTING_PREFIXES = [
    "com/celsys/",
    "com/celsys/velegacyandroidhelpers/",
]

# Method-signature patterns for subscription checks
METHOD_PATTERNS = [
    r"\.method[^\n]*\b(is[A-Z]\w*|has[A-Z]\w*|get[A-Z]\w*|check\w*|verify\w*|validate\w*|can\w*)\s*\(",
]

purchase_classes = []
for d in decoded_dir.glob("smali*"):
    if not d.is_dir():
        continue
    for f in d.rglob("*.smali"):
        rel = str(f.relative_to(decoded_dir)).replace("\\", "/")
        if any(rel.startswith(p) or p in rel for p in INTERESTING_PREFIXES):
            purchase_classes.append(f)

print(f"Found {len(purchase_classes)} classes under com/celsys/")
print()

# For each, list methods and fields
for f in purchase_classes:
    text = f.read_text(encoding="utf-8", errors="replace")
    if "purchase" not in text.lower() and "subscription" not in text.lower() and "billing" not in text.lower():
        continue
    rel = str(f.relative_to(decoded_dir))
    methods = re.findall(r"^\.method\s+(?:public\s+|private\s+|protected\s+|static\s+|final\s+|abstract\s+|synthetic\s+)*(\w+)\s*\(([^)]*)\)([^\n]*)", text, re.MULTILINE)
    fields = re.findall(r"^\.field\s+(?:public\s+|private\s+|protected\s+|static\s+|final\s+)*(\w+):([^\s]+)", text, re.MULTILINE)

    if methods or fields:
        print(f"── {rel}")
        if fields:
            print(f"   Fields: {', '.join(f'{n}:{t}' for n,t in fields[:10])}")
        if methods:
            print(f"   Methods ({len(methods)}):")
            for name, args, ret in methods[:20]:
                ret = ret.strip()
                print(f"      {name}({args}){ret}")
        print()
```
